```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import cross_val_score, train_test_split
```

```python
# Downloading data set
import requests

def download(url, filename):
    response = requests.get(url)
    if response.status_code == 200:
        with open(filename, "wb") as f:
            f.write(response.content)
        print(f"Downloaded {filename}")
```

```python
filepath = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DA0101EN-Coursera/medical_insurance_dataset.csv'
```

```python
download(filepath, "insurance.csv")
file_name="insurance.csv"
```

```python
df=pd.read_csv(file_name)
```

```python
df=pd.read_csv(file_name,header=None)
df.head(10)
```

```python
header=["age","gender","bmi","no_of_children","smoker","region","charges"]
df.columns=header
```

```python
df.replace('?',np.nan, inplace=True)
```

```python
df.info()
```

```python
is_smoker=df['smoker'].value_counts().idxmax()
df['smoker'].replace(np.nan,is_smoker,inplace=True)
age_mean=df['age'].astype(float).mean(axis=0)
df['age'].replace(np.nan,age_mean,inplace=True)
df[['smoker','age']]=df[['smoker','age']].astype(int)
df.info()
```

```python
df[['charges']]=np.round(df[['charges']],2)
```

```python
df.info()
```

```python
df.head()
```

```python
# EDA
sns.regplot(data=df,x='bmi',y='charges',line_kws={"color":"red"})
plt.ylim(0,)
```

```python
sns.boxplot(data=df,x='smoker',y='charges')
```

```python
df.corr()
```

```python
from sklearn.linear_model._glm.glm import LinearModelLoss
# MODEL DEVELOPMENT
X=df[['smoker']]
Y=df[['charges']]
lm=LinearRegression()
lm.fit(X,Y)
lm.score(X,Y)
```

```python
Z = df[["age", "gender", "bmi", "no_of_children", "smoker", "region"]]
lm.fit(Z,Y)
lm.score(Z,Y)
```

```python
Input=[('scale',StandardScaler()),('polynomial',PolynomialFeatures(include_bias=False)),('model',LinearRegression())]
pipe=Pipeline(Input)
Z=Z.astype(float)
pipe.fit(Z,Y)
ypipe=pipe.predict(Z)
r2_score(Y,ypipe)
```

```python
x_train, x_test, y_train, y_test = train_test_split(Z, Y, test_size=0.2, random_state=1)
```

```python
Ridgemodel=Ridge(alpha=0.1)
Ridgemodel.fit(x_train,y_train)
yhat=Ridgemodel.predict(x_test)
r2_score(y_test,yhat)
```

```python
pr=PolynomialFeatures(degree=2)
x_train_pr=pr.fit_transform(x_train)
x_test_pr=pr.transform(x_test)
Ridgemodel.fit(x_train_pr,y_train)
yhat_pr=Ridgemodel.predict(x_test_pr)
r2_score(y_test,yhat_pr)
```
